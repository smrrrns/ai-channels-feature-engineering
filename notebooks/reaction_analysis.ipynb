# Анализ реакций для ТГ каналов

```python
import pandas as pd
import numpy as np
from IPython.display import display, HTML
import warnings
warnings.filterwarnings('ignore')

# Настройка отображения pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
```

## Загружаем данные

```python
file_path = "../data/dataset_raw.csv" 
df = pd.read_csv(file_path)

df.head()
```

## Посмотрим на количество каждой реакции для каждого канала

```python
reaction_columns = [col for col in df.columns if col.startswith('reaction_')]

# Группируем по каналу и суммируем все реакции
channel_reactions = df.groupby('channel_title')[reaction_columns].sum()
```

### Функция для вывода

```python
def display_channel_reactions(channel_reactions_df):
    for channel in channel_reactions_df.index:
        print(f"\n{'='*60}")
        print(f"КАНАЛ: {channel}")
        print(f"{'='*60}")
        
        # Получаем все реакции для этого канала
        reactions = channel_reactions_df.loc[channel]
        non_zero_reactions = reactions[reactions > 0].sort_values(ascending=False)
      
        for reaction_col, count in non_zero_reactions.items():
            emoji = reaction_col.replace('reaction_', '')
            print(f"   {emoji}: {int(count)}")
            
        # Статистика по каналу
        total = non_zero_reactions.sum()
        print(f"\n {int(total)} реакций")
        print(f"   {len(non_zero_reactions)} типов")
```

### Выводим результаты

```python
display_channel_reactions(channel_reactions)
```
